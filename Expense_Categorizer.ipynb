# AI Personal Expense Categorizer

Practical tool jo tumhari expenses ko automatically category mein daal deta hai.

**Categories:** Food, Travel, Shopping, Bills, Entertainment, Health, Education, Other

Expense line daalo → Category + Amount nikal ke summary bhi dega.

```
from collections import defaultdict
import re

CATEGORIES = {
    "Food": ["swiggy", "zomato", "restaurant", "cafe", "food", "lunch", "dinner", "breakfast", "pizza", "burger", "chai", "coffee", "mcdonald", "kfc", "dominos", "biryani", "thali"],
    "Travel": ["uber", "ola", "rapido", "metro", "bus", "train", "flight", "petrol", "fuel", "taxi", "cab", "irctc", "indigo", "spicejet", "auto"],
    "Shopping": ["amazon", "flipkart", "myntra", "ajio", "meesho", "shopping", "clothes", "shoes", "mall", "store", "nykaa"],
    "Bills": ["electricity", "water", "gas", "internet", "wifi", "mobile", "recharge", "bill", "rent", "emi", "loan", "broadband"],
    "Entertainment": ["netflix", "prime", "hotstar", "spotify", "movie", "cinema", "pvr", "inox", "game", "steam", "youtube"],
    "Health": ["pharmacy", "medicine", "hospital", "doctor", "clinic", "apollo", "medplus", "gym", "fitness", "protein"],
    "Education": ["course", "udemy", "coursera", "book", "tuition", "college", "school", "exam", "byju"],
    "Other": []
}

def categorize(text):
    text = text.lower()
    for cat, keywords in CATEGORIES.items():
        for kw in keywords:
            if kw in text:
                return cat
    return "Other"

def extract_amount(text):
    match = re.search(r'(?:rs\.?|₹|inr)?\s*(\d+(?:\.\d{1,2})?)', text.lower())
    if match:
        return float(match.group(1))
    return 0.0

print("Expense engine ready ✅")
```

### Expenses daalo (ek ek karke ya examples se start karo)

```
# Demo examples pehle chala ke dekh lo
demo_expenses = [
    "Swiggy order 420",
    "Uber to office 185",
    "Amazon shopping 2499",
    "Netflix monthly 649",
    "Petrol 500",
    "Apollo pharmacy 320",
    "Electricity bill 1450",
    "Dominos pizza 599"
]

expenses = []
print("Demo run:\n")
for line in demo_expenses:
    cat = categorize(line)
    amt = extract_amount(line)
    expenses.append((line, cat, amt))
    print(f"{line:30} → {cat:15} ₹{amt}")
```

```
# Summary
print("\n" + "="*40)
print("SUMMARY")
print("="*40)
totals = defaultdict(float)
for _, cat, amt in expenses:
    totals[cat] += amt

for cat, total in sorted(totals.items(), key=lambda x: -x[1]):
    print(f"{cat:15} : ₹{total:.2f}")
print("-"*40)
print(f"{'TOTAL':15} : ₹{sum(totals.values()):.2f}")
```

### Apni expenses daalo

```
print("Expense line daalo (example: 'Zomato 350' ya 'Ola airport 650')")
print("Type 'done' jab khatam ho jaye\n")

while True:
    line = input("> ").strip()
    if line.lower() in ['done', 'quit', 'exit', '']:
        break
    cat = categorize(line)
    amt = extract_amount(line)
    expenses.append((line, cat, amt))
    print(f"  → {cat} | ₹{amt}")

print("\nUpdated Summary:")
totals = defaultdict(float)
for _, cat, amt in expenses:
    totals[cat] += amt
for cat, total in sorted(totals.items(), key=lambda x: -x[1]):
    print(f"{cat:15} : ₹{total:.2f}")
print(f"{'TOTAL':15} : ₹{sum(totals.values()):.2f}")
```
